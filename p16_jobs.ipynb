### SQL Table Creation Scripts

#### 1. `arxiv` Table
```sql
CREATE TABLE arxiv (
    Identifier VARCHAR(50) PRIMARY KEY,
    Date DATE,
    Title TEXT,
    Authors TEXT,
    Categories VARCHAR(255),
    Comments VARCHAR(255),
    Abstract TEXT
);
```

#### 2. `jobs` Table
```sql
CREATE TABLE jobs (
    date_posted DATE,
    title VARCHAR(255),
    company VARCHAR(255),
    location VARCHAR(255),
    num_applicants VARCHAR(255),
    skills_and_requirements TEXT,
    job_url TEXT,
    PRIMARY KEY (job_url)
);
```

#### 3. `monthly_submission` Table
```sql
CREATE TABLE monthly_submission (
    Date DATE PRIMARY KEY,
    submissions INT,
    historical_delta INT
);
```

#### 4. `NumberSubmissions` Table
```sql
CREATE TABLE NumberSubmissions (
    Date DATE,
    Num INT,
    Archive VARCHAR(50),
    PRIMARY KEY (Date, Archive)
);
```

#### 5. `trends` Table
```sql
CREATE TABLE trends (
    Date DATE PRIMARY KEY,
    AI_Value INT,
    ML_Value INT
);
```

### Notes:
- **Primary Keys**: For `arxiv`, `jobs`, and `trends`, unique identifiers (`Identifier`, `job_url`, and `Date` respectively). For `NumberSubmissions`, both `Date` and `Archive` are used as a composite primary key that you will not have multiple entries for the same archive on the same date. Adjust these primary key choices based on your actual data uniqueness and requirements.
- **Data Types**: 
  - `VARCHAR` is used for text fields that are expected to have a reasonable maximum length.
  - `TEXT` is used for fields like `Abstract` and `skills_and_requirements` that might contain longer form text.
  - `INT` is used for numeric fields without decimals.
  - `DATE` is used for date fields, assuming the format in your CSVs is compatible.

Depending on the SQL database system you use (e.g., PostgreSQL, MySQL, SQL Server), you might need to adjust data types and syntax slightly (e.g., `TEXT` might not be available in some SQL flavors, and you would use `VARCHAR(MAX)` instead).

````python
import psycopg2
from psycopg2 import sql

# Define database connection parameters
HOST = "localhost"
DATABASE = "aimltrends"
USER = "postgres"
PASS = "1231236"


def connect_db():
    # Connection to your PostgreSQL database
    return psycopg2.connect(
        host=HOST,
        database=DATABASE,
        user=USER,
        password=PASS
    )

# Function to create tables in the PostgreSQL database


def create_tables():
    # Connect to the PostgreSQL database
    with connect_db() as conn:
        with conn.cursor() as cursor:
            # List of SQL commands to execute
            commands = [
                sql.SQL('''
                CREATE TABLE IF NOT EXISTS arxiv (
                    Identifier TEXT PRIMARY KEY,
                    Date TEXT,
                    Title TEXT,
                    Authors TEXT,
                    Categories TEXT,
                    Comments TEXT,
                    Abstract TEXT
                )
                '''),
                sql.SQL('''
                CREATE TABLE IF NOT EXISTS jobs (
                    date_posted TEXT,
                    title TEXT,
                    company TEXT,
                    location TEXT,
                    num_applicants INTEGER,
                    skills_and_requirements TEXT,
                    job_id TEXT,
                    job_url TEXT PRIMARY KEY
                )
                '''),
                sql.SQL('''
                CREATE TABLE IF NOT EXISTS monthly_submission (
                    Date TEXT PRIMARY KEY,
                    submissions INTEGER,
                    historical_delta INTEGER
                )
                '''),
                sql.SQL('''
                CREATE TABLE IF NOT EXISTS NumberSubmissions (
                    Date TEXT,
                    Num INTEGER,
                    Archive TEXT,
                    PRIMARY KEY (Date, Archive)
                )
                '''),
                sql.SQL('''
                CREATE TABLE IF NOT EXISTS trends (
                    Date TEXT PRIMARY KEY,
                    AI_Value INTEGER,
                    ML_Value INTEGER
                )
                ''')
            ]

            # Execute each command
            for command in commands:
                cursor.execute(command)

            # Commit the changes
            conn.commit()
            print("Tables created successfully!")


# Call the function to create tables
create_tables()
````

Checking Duplicates & Drop them

````python
import pandas as pd


def count_duplicates(dataframe):
    # Count duplicates across the entire DataFrame
    # `keep=False` marks all duplicates as True, including the first occurrence
    duplicate_rows = dataframe.duplicated(keep=False)
    # Sum up the True values which represent duplicate rows
    count = duplicate_rows.sum()

    return count


print("arxiv: ", count_duplicates(arxiv))
print("jobs: ", count_duplicates(jobs))
print("monthly_submission: ", count_duplicates(monthly_subsmission))
print("NumberSubmissions: ", count_duplicates(NumberSubmissions))
print("trends: ", count_duplicates(trends))
````

````python
import pandas as pd
from sqlalchemy import create_engine
import psycopg2
from psycopg2 import sql


def create_engine_db():
    """Create and return an SQLAlchemy engine."""
    connection_string = f'postgresql://{USER}:{PASS}@{HOST}/{DATABASE}'
    return create_engine(connection_string)


def connect_db():
    """Connect to the PostgreSQL database."""
    return psycopg2.connect(
        host=HOST,
        database=DATABASE,
        user=USER,
        password=PASS
    )


def main():
    engine = create_engine_db()

    try:
        # Insert data into the 'arxiv' table
        arxiv.to_sql('arxiv', engine, if_exists='replace',
                     index=False, method='multi', schema='public')

        # Insert data into the 'jobs' table
        jobs.to_sql('jobs', engine, if_exists='replace',
                    index=False, method='multi', schema='public')

        # Insert data into the 'monthly_submission' table
        monthly_subsmission.to_sql('monthly_submission', engine,
                                   if_exists='replace', index=False, method='multi', schema='public')

        # Insert data into the 'NumberSubmissions' table
        NumberSubmissions.to_sql('NumberSubmissions', engine, if_exists='replace',
                                 index=False, method='multi', schema='public')

        # Insert data into the 'trends' table
        trends.to_sql('trends', engine, if_exists='replace',
                      index=False, method='multi', schema='public')

        print("Data inserted successfully!")
    except Exception as e:
        print(f"An error occurred: {str(e)}")
    finally:
        engine.dispose()


if __name__ == '__main__':
    main()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.graphics.tsaplots import plot_acf
import psycopg2


def process_time_series(table_name, date_col, value_col, arima_order=(1, 1, 1), convert_numeric=False):
    # Connect to the PostgreSQL database
    conn = psycopg2.connect(
        dbname=DATABASE,
        user=USER,
        password=PASS,
        host=HOST,
    )

    # Load data from PostgreSQL table
    query = f"SELECT * FROM {table_name}"
    df = pd.read_sql_query(query, conn)
    print(df)
    # Convert date to datetime and set as index
    df[date_col] = pd.to_datetime(df[date_col])
    df.set_index(date_col, inplace=True)

    # Convert value_col to numeric if specified
    if convert_numeric:
        df[value_col] = pd.to_numeric(df[value_col], errors='coerce')

    # Handle missing values
    df = df.fillna(method='ffill')

    # Plot the data
    df[value_col].plot(title=f'Time Series Plot - {value_col}')
    plt.show()

    # Fit ARIMA model
    model = SARIMAX(df[value_col], order=arima_order,
                    seasonal_order=(0, 0, 0, 0))
    result = model.fit()

    # After fitting the model with SARIMAX as previously done
    residuals = result.resid

    # Plot residuals
    plt.figure(figsize=(10, 5))
    plt.plot(residuals)
    plt.title('Residuals from ARIMA model')
    plt.xlabel('Date')
    plt.ylabel('Residual')
    plt.show()

    # Check for no autocorrelation in residuals
    plt.figure(figsize=(10, 5))
    plot_acf(residuals, lags=20, title="ACF of Residuals")
    plt.show()

    # Forecast the next 12 months
    forecast = result.get_forecast(steps=12)
    forecast_df = forecast.conf_int()
    forecast_df['forecast'] = result.predict(
        start=forecast_df.index[0], end=forecast_df.index[-1])

    # Plot the forecast
    plt.figure(figsize=(10, 5))
    plt.plot(df[value_col], label='Actual')
    plt.plot(forecast_df['forecast'], label='Forecast', color='red')
    plt.fill_between(
        forecast_df.index, forecast_df.iloc[:, 0], forecast_df.iloc[:, 1], color='pink')
    plt.title('Forecast vs Actuals')
    plt.legend()
    plt.show()

    # Close the database connection
    conn.close()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
import psycopg2
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf


def process_time_series(table_name, date_col, value_col, arima_order=(1, 1, 1), convert_numeric=True, dbname=None, user=None, password=None, host=None):
    # Connect to the PostgreSQL database
    conn = psycopg2.connect(
        dbname=dbname,
        user=user,
        password=password,
        host=host
    )

    # Load data from SQL table
    query = f"SELECT {date_col}, {value_col} FROM {table_name}"
    df = pd.read_sql_query(query, conn)
    print(df)
    # Convert date to datetime and set as index
    df[date_col] = pd.to_datetime(df[date_col])
    df.set_index(date_col, inplace=True)

    # Convert value_col to numeric if specified
    if convert_numeric:
        df[value_col] = pd.to_numeric(df[value_col], errors='coerce')

    # Handle missing values
    df.dropna(inplace=True)  # Ensure all NA/NaN are dropped after conversion

    # Group by date and count number of applicants
    df_grouped = df.groupby(pd.Grouper(freq='D')).sum().reset_index()

    # Check if there's enough data to analyze
    if len(df_grouped) < 30:
        print(
            f"Not enough data to perform time series analysis. Only {len(df_grouped)} data points available.")
        return

    # Plot the data
    plt.figure(figsize=(10, 5))
    df_grouped[value_col].plot(title=f'Time Series Plot - {value_col}')
    plt.show()

    # Plot ACF and PACF
    plt.figure(figsize=(10, 5))
    plot_acf(df_grouped[value_col], lags=20, title="ACF Plot")
    plt.show()

    plt.figure(figsize=(10, 5))
    plot_pacf(df_grouped[value_col], lags=20, title="PACF Plot")
    plt.show()

    # Fit ARIMA model
    model = SARIMAX(df_grouped[value_col], order=arima_order,
                    seasonal_order=(0, 0, 0, 0))
    result = model.fit(disp=False)

    # Forecast the next 12 months
    forecast = result.get_forecast(steps=12)
    forecast_df = forecast.conf_int()
    forecast_df['forecast'] = result.predict(
        start=forecast_df.index[0], end=forecast_df.index[-1])

    # Plot the forecast
    plt.figure(figsize=(10, 5))
    plt.plot(df_grouped[value_col], label='Actual')
    plt.plot(forecast_df['forecast'], label='Forecast', color='red')
    plt.fill_between(
        forecast_df.index, forecast_df.iloc[:, 0], forecast_df.iloc[:, 1], color='pink')
    plt.title('Forecast vs Actuals')
    plt.legend()
    plt.show()

    # Close the database connection
    conn.close()


# Example usage:
process_time_series('jobs', 'date_posted', 'num_applicants', dbname=DATABASE,
                    user=USER, password=PASS, host=HOST)
````
